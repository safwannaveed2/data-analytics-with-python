```python
class Student:
    def __init__(self, name, age, score):
        self.name = name
        self.age = age
        self.score = score

    def get_grade(self): # Class Method 
        if self.score >= 90:
            return "A"
        elif self.score >= 80: 
            return "B"
        else: 
            return "C"
        
    def is_passing(self):
        return self.score >= 50
    

s1 = Student("Misha", 25, 90)
s2 = Student("Aftab",15,40)

print(s1.is_passing())
print(s2.is_passing())
```

```python
class Student:
    def __init__(self, name, age, score):
        self.name = name
        self.age = age
        self.score = score

    def get_grade(self): # Class Method 
        if self.score >= 90:
            return "A"
        elif self.score >= 80: 
            return "B"
        else: 
            return "C"
        
    def is_passing(self):
        return self.score >= 50
    
    def print_method(self):
        return (f'{self.name} | {self.age} | {self.score}')

    def __str__(self):
        return (f'{self.name} | {self.age} | {self.score}')

s1 = Student("Misha", 25, 90)
s2 = Student("Aftab",15,40)

print(s1)
print(s1.is_passing())

# print(s1.print_method())
```
