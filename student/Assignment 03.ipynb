#### 1. Write a program that accepts a list from user and print the alternate element of list.

```python
numbers = input("Enter list elements separated by space: ").split()

print("Alternate elements:")

for i in range(0, len(numbers), 2):
    print(numbers[i])
```

#### 2. Write a program that accepts a list from user. Your program should reverse the content of list anddisplay it. Do not use reverse() method.

```python
numbers = input("Enter list elements separated by space: ").split()

reversed_list = []

for i in range(len(numbers) - 1, -1, -1):
    reversed_list.append(numbers[i])

print("Reversed list:", reversed_list)
```

#### 3. Find and display the largest number of a list without using built-in function max(). Your programshould ask the user to input values in list from keyboard.

```python
numbers = input("Enter numbers separated by space: ").split()

numbers = [int(x) for x in numbers]

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest number:", largest)
```

#### 4. Write a program that rotates the element of a list so that the element at the first index moves to thesecond index, the element in the second index moves to the third index, etc., and the element in the last index moves to the first index.
 

```python
numbers = input("Enter list elements separated by space: ").split()

rotated = [numbers[-1]]

for i in range(len(numbers) - 1):
    rotated.append(numbers[i])

print("Rotated list:", rotated)
```

#### 5. Write a program that input a string and ask user to delete a given word from a string.

```python
sentence = input("Enter a sentence: ")
word = input("Enter the word you want to delete: ")

words = sentence.split()
new_sentence = []

for w in words:
    if w != word:
        new_sentence.append(w)

print("New sentence:", " ".join(new_sentence))
```

#### 6. Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It should print the date in the form March 12, 2021.

```python
date = input("Enter date (mm/dd/yyyy): ")

month, day, year = date.split("/")

months = {
    "01": "January",
    "02": "February",
    "03": "March",
    "04": "April",
    "05": "May",
    "06": "June",
    "07": "July",
    "08": "August",
    "09": "September",
    "10": "October",
    "11": "November",
    "12": "December"
}

print(months[month], day + ",", year)
```

#### 7. Write a program with a function that accepts a string from keyboard and create a new string afterconverting character of each word capitalized. For instance, if the sentence is "stop and smell the roses." the output should be "Stop And Smell The Roses" 

```python
def capitalize_words(sentence):
    words = sentence.split()
    new_sentence = []

    for word in words:
        new_sentence.append(word.capitalize())

    return " ".join(new_sentence)


sentence = input("Enter a sentence: ")

result = capitalize_words(sentence)

print(result)
```

#### 8. Find the sum of each row of matrix of size m x n. For example for the following matrix output will be like this :first row 2 7 11 12 second row 5 2 9 15 third row 8 3 10 42  Sum of row 1 = 32  Sum of row 2 = 31  Sum of row 3 = 63

```python
matrix = [
    [2, 7, 11, 12],
    [5, 2, 9, 15],
    [8, 3, 10, 42]
]

for i in range(len(matrix)):
    total = 0

    for j in range(len(matrix[i])):
        total = total + matrix[i][j]

    print("Sum of row", i + 1, "=", total)
```

#### 9. Write a program to add two matrices of size n x m.

```python
matrix1 = [
    [1, 2],
    [3, 4]
]

matrix2 = [
    [5, 6],
    [7, 8]
]

result = []

for i in range(len(matrix1)):
    row = []

    for j in range(len(matrix1[i])):
        row.append(matrix1[i][j] + matrix2[i][j])

    result.append(row)

print("Result:")

for row in result:
    print(row)
```

#### 10. Write a program to multiply two matrices 

```python
matrix1 = [
    [1, 2],
    [3, 4]
]

matrix2 = [
    [5, 6],
    [7, 8]
]

result = [
    [0, 0],
    [0, 0]
]

for i in range(len(matrix1)):
    for j in range(len(matrix2[0])):
        for k in range(len(matrix2)):
            result[i][j] = result[i][j] + matrix1[i][k] * matrix2[k][j]

print("Result:")

for row in result:
    print(row)
```
